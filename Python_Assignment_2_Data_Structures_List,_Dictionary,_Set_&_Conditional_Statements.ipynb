List (Creation, Modification and Access)

1. List Creation:

* A list named age_list is created to store five integer values. Here the integer elements are 24,25,26,27,28.
* Also created a name_list to store string elements



```python
age_list=[24,25,26,27,28]
print(age_list)
```

```python
name_list=["kripa","christy","aleena","jincy","alan"]
print(name_list)
```

2. List Operations / Modifications:

a. append() method is used to add a new element to the end of a list.Here, the string "Yazhini" is added to the end of the list called name_list.

```python
name_list.append("Yazhini")
print(name_list)
```

b. here insert() method is used to add an element using its index number.30 is added to the second position.

```python
age_list.insert(2,30)
print(age_list)
```

c. remove() method is used to remove a specific value in a list , here using remove method removed the string "Yazhini".

```python
name_list.remove("Yazhini")
print(name_list)
```

d. pop()- used to remove an item using its index ,here we not give any index value inside pop, so without giving index value pop removes the last element.

```python
age_list.pop()
print(age_list)
```

e. extend() is used to add mutiple values , here we add [29, 30, 26] to the age_list

```python
age_list.extend([29, 30, 26])
print(age_list)
```

f. age_list is arranged in descending order, Here, the highest age is 30 and the lowest age is 24 and we use .sort(reverse=True) method to sort this list into descending order.

```python
age_list.sort(reverse=True)
print(age_list)
```

g. here we find the maximum age, minimum age and sum of all ages
* max() function is used to find the maximum age value
* min() function is used to find the minimum age value
* sum() function is used to find the total of all age values


```python
max_age=max(age_list)
print(max_age)


min_age=min(age_list)
print(min_age)

sum_of_all_ages=sum(age_list)
print(sum_of_all_ages)
```

3. Accessing List Elements:

a.The index of the first element is always 0. So the first element in the name_list is "kripa".Therefore, name_list[0] is used to access the first element of the list.

```python
print(name_list[0])
```

b.For access last element in the name_list we should have use negative indexing.The index -1 always represents the last element of the list. In this name_list last element ia "alan"

```python
print(name_list[-1])
```

c. Here we want index 2 to 4 elements, 2 is the starting index and 5 is the stopping index. Python does not include the stopping index, so index 5 is not included. So we get values from index 2 to 4

```python
print(name_list[2:5])
```

d. Here we can use reverse() function to reverse this name_list

```python
name_list.reverse()
print(name_list)
```

Dictionary (Creation, Modification and Access):

a. Dictionary named student_marks is created to store the names and marks of five students.

```python
student_marks={"Sreya":45,"neethu":53,"kiran":62,"ayana":39,"meera":76}
print(student_marks)
```

b. Dictionary value can be accessed by using its key, here "neethu" is the key .Using "neethu" we can retrive her marks.

```python
print(student_marks["neethu"])
```

c. A new key "Janani" is added and assigned the value 80

```python
student_marks["Janani"] = 80
print(student_marks)
```

d.value be updated by using its existing key. After updating, kiran's entry in the dictionary becomes "kiran":85 , kiran's marks changes from 62 to 85

```python
student_marks["kiran"]=85
print(student_marks)

```

e. keys() displays the names of all students because the student names are the keys

* values() displays the marks of all students because the marks are the values in the dictionary

* items() displays both keys and values here both students and marks

```python
student_marks.keys()
print(student_marks.keys())


student_marks.values()
print(student_marks.values())

student_marks.items()
print(student_marks.items())
```

Performance Category Program:

1.Here the input() function is used to prompt the user to enter a score

```python
score = int(input("Enter your score (0-10): "))
print(score)
```

2&3.  Here the input score given is 6 , the program checks the score using given conditions. Here 6 is not greater than 7 , so first condition not satisfying . '6' is between 4 and 7 so the second condition is true

```python
if score > 7:
    print("Above Average")
elif score >= 4:
    print("Average")
else:
    print("Below Average")
```

4.The additional step is to display a suitable message or feedback based on the performance category.Here our score is '6' and its an average score so print Good effort!.

```python
if score > 7:
    performance = "Above Average"
    message = "Excellent work!"

elif score >= 4:
    performance = "Average"
    message = "Good effort!"

else:
    performance = "Below Average"
    message = "Keep practicing!"

print(performance + ": " + message)
```
